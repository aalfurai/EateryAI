## Load Data

```python
import json
import pandas as pd

with open('../restaurants.menu_item_variations.json', 'r') as file:
    item_list = json.load(file)
file.close()
```

## Select all Items for Specific Restaurants: Chick-fil-A and Chipotle

```python
chick_fil_a_items = []
chipotle_items = []

for item in item_list:
    if item.get("restaurant_name") == "Chick-fil-A":
        chick_fil_a_items.append(item)

    elif item.get("restaurant_name") == "Chipotle":
        chipotle_items.append(item)

print(f"Number of Chick-fil-A items: {len(chick_fil_a_items)}")
print(f"Number of Chipotle items: {len(chipotle_items)}")

# making dataframes just to visualize easier
df_cfa = pd.DataFrame(chick_fil_a_items)
df_chipotle = pd.DataFrame(chipotle_items)
```

## Assign Meal Categories (Entree, Side, Drink, Dessert, Add-On)
Note: this is hard-coded just to have categories to work with.

### Chick-fil-A
Concern: What to do about something like salad dressings? How would we account for a combination that only makes sense in a certain context?


```python
# add a column for meal categories
df_cfa['meal_category'] = 'temp'

# assign meal categories
# note: hard-coded assignment for this one restaurant

entree_categories = ['Classic Chicken', 'Breakfast', 'Cool Wraps', 'Salads', ]
df_cfa.loc[df_cfa['category'].isin(entree_categories), 'meal_category'] = 'Entree'

df_cfa.loc[df_cfa['category'] == 'Sides', 'meal_category'] = 'Side'

df_cfa.loc[df_cfa['category'] == 'Beverages', 'meal_category'] = 'Drink'

df_cfa.loc[df_cfa['category'] == 'Desserts', 'meal_category'] = 'Dessert'

addon_categories = ['Salad Dressings', 'Sauces']
df_cfa.loc[df_cfa['category'].isin(addon_categories), 'meal_category'] = 'Add-On'

# assigning some categories manually
df_cfa.loc[90, 'meal_category'] = 'Side'

kid_mask = df_cfa['category'] == "Kid's Meal"
df_cfa.loc[kid_mask & df_cfa['menu_item_name'].str.contains('Meal', case=False, na=False),'meal_category'] = 'Entree'
df_cfa.loc[kid_mask & df_cfa['menu_item_name'].str.contains('Milk', case=False, na=False),'meal_category'] = 'Drink'
df_cfa.loc[kid_mask & ~df_cfa['menu_item_name'].str.contains('Meal|Milk', case=False, na=False),'meal_category'] = 'Side'
```

```python
# check that all items got assigned a category
display(df_cfa.loc[df_cfa['meal_category'] == 'temp'])
```

```python
# sort menu items for CFA by meal category
# placeholder: storing menu items just by their ID

cfa_entrees = list(df_cfa.loc[df_cfa['meal_category'] == 'Entree', 'menu_item_id'])
cfa_sides = list(df_cfa.loc[df_cfa['meal_category'] == 'Side', 'menu_item_id'])
cfa_drinks = list(df_cfa.loc[df_cfa['meal_category'] == 'Drink', 'menu_item_id'])
cfa_desserts = list(df_cfa.loc[df_cfa['meal_category'] == 'Dessert', 'menu_item_id'])
cfa_addons = list(df_cfa.loc[df_cfa['meal_category'] == 'Add-On', 'menu_item_id'])

print(f"Count of Entrees: {len(cfa_entrees)}")
print(f"Count of Sides: {len(cfa_sides)}")
print(f"Count of Drinks: {len(cfa_drinks)}")
print(f"Count of Desserts: {len(cfa_desserts)}")
print(f"Count of Add-Ons: {len(cfa_addons)}")
```

## Meal Combination Exploration

Information to calculate when storing meals:
- calories
- protein
- price

```python
import itertools
```

```python
# forcing the combo to at least have Entree, Side, and Drink
def compute_cfa_combos(addons=False, desssert=False):
    categories = [cfa_entrees, cfa_sides, cfa_drinks]
    
    if addons:
        categories.append(cfa_addons)
    
    if desssert:
        categories.append(cfa_desserts)
    
    combinations = itertools.product(*categories)
    return combinations     # returns an iterator object
```

```python
iter_cfa_combo_basic = compute_cfa_combos()
list_cfa_combo_basic = list(iter_cfa_combo_basic)
print(f"Number of total combinations for Entree + Side + Drink combo: {len(list_cfa_combo_basic)}")

iter_cfa_combo_extra = compute_cfa_combos(addons=True, desssert=True)
list_cfa_combo_extra = list(iter_cfa_combo_extra)
print(f"Number of total combinations for Entree, Side, Drink, Dessert and Add-On combo: {len(list_cfa_combo_extra)}")
```
